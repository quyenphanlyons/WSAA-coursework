Write a program in python that will read a file from a repository, 

The program should then replace all the instances of the text "Andrew" with your name. 

The program should then commit those changes and push the file back to the repository (You will need authorisation to do this).

```python
import os
from git import Repo

REPO_URL = "git@github.com:quyenphanlyons/Other-files-rep.git"
LOCAL_DIR = "repo_clone"
FILE_PATH = "The-humans.txt"

# Create a repo clone of the repository 'Other-files-rep' in 'assignements' folder
def clone_repo():
    # Clone the repository if it doesn't already exist
    if not os.path.exists(LOCAL_DIR):
        Repo.clone_from(REPO_URL, LOCAL_DIR)
    else:
        print('Repository already cloned.')

# Modify the text file 'The-humans'
def modify_file():
    #"""Replace 'Andrew' with 'Quyen' in the file"""
    path = os.path.join(LOCAL_DIR, FILE_PATH)

    with open(path, 'r', encoding='utf-8') as file:
        content = file.read()

    updated_content = content.replace('Andrew', 'Quyen')

    with open(path, 'w', encoding='utf-8') as file:
        file.write(updated_content)

    print('File updated successfully.')

# Commit and push changes to Github
def commit_and_push():
    #"""Commit and push changes to GitHub"""
    repo = Repo(LOCAL_DIR)

    # Change the remote URL of the local repository 
    repo.git.remote('set-url', 'origin', REPO_URL)

    # Stages all changes
    repo.git.add(A=True)

    # Commit only if changes exist
    if repo.is_dirty(untracked_files=True):
        repo.index.commit("Replaced 'Andrew' with 'Quyen' in The-humans.txt")
        print("Committed changes.")
    else:
        print("No changes to commit.")

    # Push
    origin = repo.remote(name='origin')
    origin.push()

    print("Changes pushed successfully.")


def main():
    clone_repo()
    modify_file()
    commit_and_push()


if __name__ == "__main__":
    main()
```
